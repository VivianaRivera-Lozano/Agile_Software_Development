```python
from pathlib import Path
import pandas as pd

folder = Path(
    r"C:/Users/vivir/OneDrive/Escritorio/TERMS/FALL 2026/Agile Software development"
)

# Load the CSV file
file_path = folder / "cleaned_Airbnb_NYC.csv"

df = pd.read_csv(file_path)

# Display the first five rows
print(df.head())
```

```python
#   Task: 1 Aggregate prices by neighborhood group and calculate the mean
prices_per_neighborhood = df.groupby("neighbourhood_group")["price"].mean()
print(prices_per_neighborhood)
```

```python
# Task 2: Create a filter for price with a slider

import ipywidgets as widgets
min_price, max_price =df["price"].min(), df["price"].max()
print(f"Price range: {min_price} - {max_price}")

price_slider = widgets.FloatRangeSlider(
    value=[min_price, max_price],
    min=min_price,
    max=max_price,
    description="Price Range:",
    continuous_update=False,
)

# format for slider

price_slider.layout = widgets.Layout(width="600px", margin="20px 0 20px 0")
price_slider.style.description_width = "120px"
price_slider.readout_format = ".2f"
price_slider.style.handle_color = "blue"
```
